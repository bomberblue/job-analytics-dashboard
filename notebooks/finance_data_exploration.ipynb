## Data Cleaning Pipeline - Start Here

### Common Validation/Parsing Functions

```python
import json
import pandas as pd

# Parse JSON strings into Python lists; skip if already parsed
def _parse_categories(value):
        if isinstance(value, list):
            return value
        if isinstance(value, str):
            value = value.strip()
            if not value:
                return []
            try:
                parsed = json.loads(value)
            except json.JSONDecodeError as e:
                raise ValueError(f"Malformed JSON in categories: {value[:200]}") from e
            return parsed if isinstance(parsed, list) else [parsed]
        if pd.isna(value):
            return []
        raise ValueError(f"Unsupported categories value type: {type(value).__name__}")

def validate_non_negative(chunk, columns):
    """Raise an error when a specified numeric column contains negative values."""
    missing_columns = [column for column in columns if column not in chunk.columns]
    if missing_columns:
        raise KeyError(f"Missing columns for non-negative validation: {missing_columns}")

    for column in columns:
        negative_count = (chunk[column] < 0).sum()
        if negative_count:
            raise ValueError(f"{column}: {negative_count} negative values are not allowed.")
```

### Data Cleaning

```python
import json
from pathlib import Path

source_path = Path("../data/raw/SGJobData.csv")
output_path = Path("../data/processed/SGJobData_cleaned.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
chunk_size = 100_000

expected_dtypes = {
    "metadata_isPostedOnBehalf": "bool",
    "metadata_repostCount": "int64",
    "metadata_totalNumberJobApplication": "int64",
    "metadata_totalNumberOfView": "int64",
    "minimumYearsExperience": "int64",
    "numberOfVacancies": "int64",
    "salary_maximum": "float64",
    "salary_minimum": "float64",
    "average_salary": "float64",
}
required_columns = [
    "metadata_originalPostingDate", "metadata_newPostingDate",
    "metadata_expiryDate", "metadata_jobPostId", "postedCompany_name",
]
date_columns = [
    "metadata_expiryDate", "metadata_newPostingDate", "metadata_originalPostingDate",
]


float_columns = [
    column
    for column, dtype in expected_dtypes.items()
    if dtype == "float64"
]

non_negative_columns = [
    "average_salary",
    "salary_minimum",
    "salary_maximum",
]


int_columns = [column for column, dtype in expected_dtypes.items() if dtype == "int64"]

def clean_chunk(chunk):

    missing_columns = [column for column in [*required_columns, *expected_dtypes] if column not in chunk.columns]
    if missing_columns:
        raise KeyError(f"Missing columns: {missing_columns}")

    chunk = chunk.dropna(subset=required_columns, how="all").copy()

    for column in date_columns:
        chunk[column] = pd.to_datetime(chunk[column], errors="coerce", cache=False)

    bool_column = "metadata_isPostedOnBehalf"
    parsed_bool = (
        chunk[bool_column].astype(str).str.strip().str.lower()
        .map({"true": True, "false": False, "1": True, "0": False, "yes": True, "no": False})
    )
    if parsed_bool.isna().any():
        raise ValueError(f"{bool_column}: {parsed_bool.isna().sum()} invalid values cannot be parsed to bool.")
    chunk[bool_column] = parsed_bool.astype(bool)

    for column in int_columns:
        parsed = pd.to_numeric(chunk[column], errors="coerce")
        if parsed.isna().any():
            raise ValueError(f"{column}: {parsed.isna().sum()} invalid values cannot be parsed to int64.")
        chunk[column] = parsed.astype("int64")

    for float_column in float_columns:
        parsed_float = pd.to_numeric(chunk[float_column], errors="coerce")
        if parsed_float.isna().any():
            raise ValueError(f"{float_column}: {parsed_float.isna().sum()} invalid values cannot be parsed to float64.")
        chunk[float_column] = parsed_float.astype("float64")
    
    validate_non_negative(chunk, non_negative_columns)

    dtype_check = chunk[list(expected_dtypes)].dtypes.astype(str).to_dict()
    assert dtype_check == expected_dtypes, "Type validation failed for one or more columns."
    chunk = chunk.drop(columns=["occupationId"], errors="ignore")

    chunk['metadata_expiryDate'] = pd.to_datetime(chunk['metadata_expiryDate'], errors='coerce', cache=False)
    chunk['metadata_newPostingDate'] = pd.to_datetime(chunk['metadata_newPostingDate'], errors='coerce', cache=False)
    chunk['metadata_originalPostingDate'] = pd.to_datetime(chunk['metadata_originalPostingDate'], errors='coerce', cache=False)

    chunk["categories"] = chunk["categories"].apply(_parse_categories)

    return chunk

rows_written = 0
for chunk_number, chunk in enumerate(pd.read_csv(source_path, chunksize=chunk_size)):
    cleaned_chunk = clean_chunk(chunk)
    cleaned_chunk.to_csv(output_path, mode="w" if chunk_number == 0 else "a", header=chunk_number == 0, index=False)
    rows_written += len(cleaned_chunk)
    # display(cleaned_chunk.info())

print(f"Streaming clean completed: {rows_written:,} rows written to {output_path}.")
```

```python
from pathlib import Path
import ast
import pandas as pd

def load_categories(value):
    try:
        parsed = ast.literal_eval(value)
    except (SyntaxError, ValueError) as error:
        raise ValueError(f"Invalid Python list in categories: {value[:200]}") from error

    if not isinstance(parsed, list):
        raise ValueError("categories must contain a JSON list.")
    return parsed


clean_path = Path("../data/processed/SGJobData_cleaned.csv")

for chunk in pd.read_csv(
    clean_path,
    chunksize=100_000,
    parse_dates=[
        "metadata_expiryDate",
        "metadata_newPostingDate",
        "metadata_originalPostingDate",
    ],
    converters={"categories": load_categories},
):
    with pd.option_context("display.float_format", "{:,.2f}".format):
        display(chunk.describe())
    
```
